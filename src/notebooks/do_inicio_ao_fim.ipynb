```python
# Ok, agora preciso criar os dados de treino e teste para a criação do modelo de classificação
from sklearn.metrics.pairwise import cosine_similarity

idx_
A = df.loc[2112, "embedding"]
B = df.loc[1114, "embedding"]

similarities = []
for B in df.iloc[:, -1]:
    similarities.append(cosine_similarity(X=A, Y=B).item())
```

```python
import numpy as np

n = 10
top_n_idx = np.argpartition(similarities, -n)[-n:]
argmax = np.argmax(similarities)

# print(df.iloc[0, 1])
print(df.iloc[argmax, 1])
```
